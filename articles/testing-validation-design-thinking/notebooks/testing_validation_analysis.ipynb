# Testing and Validation Analysis Notebook

This notebook accompanies the article **Testing and Validation in Design Thinking**. It provides an exploratory workflow for comparing tested concepts across desirability, feasibility, viability, responsibility, friction, residual risk, evidence quality, and stakeholder coverage.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
concepts_path = article_dir / 'data' / 'raw' / 'validation_concepts_raw.csv'
concepts = pd.read_csv(concepts_path)
concepts
```

```python
concepts['combined_risk'] = 0.50 * concepts['friction'] + 0.50 * concepts['residual_risk']
concepts['validation_value'] = (
    0.25 * concepts['desirability'] +
    0.20 * concepts['feasibility'] +
    0.20 * concepts['viability'] +
    0.20 * concepts['responsibility'] -
    0.15 * concepts['combined_risk']
)
concepts.sort_values('validation_value', ascending=False)
```
