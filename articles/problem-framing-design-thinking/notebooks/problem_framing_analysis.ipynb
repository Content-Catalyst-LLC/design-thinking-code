# Problem Framing Analysis Notebook

This notebook accompanies the article **Problem Framing in Design Thinking**. It provides an exploratory workflow for comparing candidate frames across explanatory adequacy, stakeholder coverage, opportunity value, framing risk, and evidence confidence.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
data_path = article_dir / 'data' / 'raw' / 'problem_frames_raw.csv'
df = pd.read_csv(data_path)
df
```

```python
weights = {
    'explanatory_adequacy': 0.30,
    'stakeholder_coverage': 0.25,
    'opportunity_value': 0.30,
    'framing_risk': 0.15,
}

df['frame_value'] = (
    weights['explanatory_adequacy'] * df['explanatory_adequacy'] +
    weights['stakeholder_coverage'] * df['stakeholder_coverage'] +
    weights['opportunity_value'] * df['opportunity_value'] -
    weights['framing_risk'] * df['framing_risk']
)

df.sort_values('frame_value', ascending=False)
```
