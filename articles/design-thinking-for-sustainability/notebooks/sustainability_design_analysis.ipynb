# Design Thinking for Sustainability

This notebook accompanies the article **Design Thinking for Sustainability**. It provides an exploratory workflow for comparing sustainability design concepts across usability, feasibility, ecological benefit, circularity, equity, durability, risk, evidence quality, stakeholder coverage, and lifecycle-boundary quality.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
concepts_path = article_dir / 'data' / 'raw' / 'sustainability_concepts_raw.csv'
concepts = pd.read_csv(concepts_path)
concepts
```

```python
concepts['evidence_strength'] = (
    0.35 * concepts['evidence_quality'] +
    0.30 * concepts['stakeholder_coverage'] +
    0.35 * concepts['lifecycle_boundary_quality']
)
concepts['sustainability_value'] = (
    0.16 * concepts['usability'] +
    0.16 * concepts['feasibility'] +
    0.24 * concepts['ecological_benefit'] +
    0.16 * concepts['circularity'] +
    0.14 * concepts['equity'] +
    0.08 * concepts['durability'] -
    0.06 * concepts['risk']
)
concepts['evidence_adjusted_value'] = concepts['sustainability_value'] * (0.75 + 0.25 * concepts['evidence_strength'])
concepts.sort_values('sustainability_value', ascending=False)
```
