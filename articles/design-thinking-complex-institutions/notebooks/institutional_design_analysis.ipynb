# Design Thinking for Complex Institutions

This notebook accompanies the article **Design Thinking for Complex Institutions**. It explores institutional design options, change readiness, public value, absorption capacity, sequencing need, and implementation risk.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
options = pd.read_csv(article_dir / 'data' / 'raw' / 'institutional_design_options_raw.csv')
options.head()
```

```python
options['change_readiness'] = (
    0.14 * options['desirability'] +
    0.13 * options['authority'] +
    0.12 * options['capability'] +
    0.10 * options['funding'] +
    0.10 * options['policy_fit'] +
    0.11 * options['governance_strength'] +
    0.08 * options['trust_gain'] +
    0.08 * options['burden_reduction'] -
    0.05 * options['coordination_complexity'] -
    0.05 * options['implementation_risk']
)
options[['option', 'option_type', 'change_readiness']].sort_values('change_readiness', ascending=False)
```
