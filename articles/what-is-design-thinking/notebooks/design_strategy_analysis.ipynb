# Design Strategy Analysis Notebook

This notebook accompanies the article **What Is Design Thinking?** It provides an exploratory workflow for comparing design pathways across human relevance, feasibility, learning value, and residual risk.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
data_path = article_dir / 'data' / 'raw' / 'design_pathways_raw.csv'
df = pd.read_csv(data_path)
df
```

```python
weights = {
    'human_relevance': 0.35,
    'feasibility': 0.25,
    'learning_value': 0.25,
    'residual_risk': 0.15,
}

df['design_value'] = (
    weights['human_relevance'] * df['human_relevance'] +
    weights['feasibility'] * df['feasibility'] +
    weights['learning_value'] * df['learning_value'] -
    weights['residual_risk'] * df['residual_risk']
)

df.sort_values('design_value', ascending=False)
```
