# Implementation and Scaling Analysis Notebook

This notebook accompanies the article **Implementation and Scaling in Design Thinking**. It provides an exploratory workflow for comparing implementation candidates across adoption readiness, operational fit, durability, governance readiness, equity readiness, financial sustainability, risk, context complexity, and scale sensitivity.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
portfolio_path = article_dir / 'data' / 'raw' / 'implementation_portfolio_raw.csv'
portfolio = pd.read_csv(portfolio_path)
portfolio
```

```python
portfolio['composite_risk'] = (
    0.25 * portfolio['operational_risk'] +
    0.22 * portfolio['governance_risk'] +
    0.18 * portfolio['technical_risk'] +
    0.22 * portfolio['equity_risk'] +
    0.13 * portfolio['financial_risk']
)

portfolio['implementation_value'] = (
    0.20 * portfolio['adoption_readiness'] +
    0.18 * portfolio['operational_fit'] +
    0.18 * portfolio['durability'] +
    0.15 * portfolio['governance_readiness'] +
    0.14 * portfolio['equity_readiness'] +
    0.10 * portfolio['financial_sustainability'] -
    0.05 * portfolio['composite_risk']
)

portfolio['scaled_quality_estimate'] = (
    portfolio['implementation_value'] - portfolio['scale_sensitivity'] * portfolio['context_complexity']
)

portfolio.sort_values('implementation_value', ascending=False)
```
