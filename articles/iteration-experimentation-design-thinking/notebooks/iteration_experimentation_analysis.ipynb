# Iteration and Experimentation Analysis Notebook

This notebook accompanies the article **Iteration and Experimentation in Design Thinking**. It provides an exploratory workflow for comparing experiments across learning gain, update flexibility, expected improvement, residual risk, and evidence confidence.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
data_path = article_dir / 'data' / 'raw' / 'experiments_raw.csv'
df = pd.read_csv(data_path)
df
```

```python
weights = {
    'learning_gain': 0.35,
    'update_flexibility': 0.25,
    'expected_improvement': 0.25,
    'residual_risk': 0.15,
}

df['experiment_value'] = (
    weights['learning_gain'] * df['learning_gain'] +
    weights['update_flexibility'] * df['update_flexibility'] +
    weights['expected_improvement'] * df['expected_improvement'] -
    weights['residual_risk'] * df['residual_risk']
)

df.sort_values('experiment_value', ascending=False)
```
