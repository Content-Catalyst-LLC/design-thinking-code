# Prototype Portfolio Analysis Notebook

This notebook accompanies the article **Prototyping in Design Thinking**. It provides an exploratory workflow for comparing prototypes across learning gain, feasibility signal, user response, equity value, implementation relevance, and composite risk.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
portfolio_path = article_dir / 'data' / 'raw' / 'prototype_portfolio_raw.csv'
portfolio = pd.read_csv(portfolio_path)
portfolio
```

```python
portfolio['composite_risk'] = (
    0.30 * portfolio['ethical_risk'] +
    0.30 * portfolio['operational_risk'] +
    0.20 * portfolio['technical_risk'] +
    0.20 * portfolio['scaling_risk']
)

portfolio['prototype_value'] = (
    0.25 * portfolio['learning_gain'] +
    0.18 * portfolio['feasibility_signal'] +
    0.20 * portfolio['user_response'] +
    0.15 * portfolio['equity_value'] +
    0.12 * portfolio['implementation_relevance'] -
    0.10 * portfolio['composite_risk']
)

portfolio.sort_values('prototype_value', ascending=False)
```
