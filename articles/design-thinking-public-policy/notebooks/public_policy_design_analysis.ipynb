# Design Thinking in Public Policy

This notebook accompanies the article **Design Thinking in Public Policy**. It provides an exploratory workflow for comparing public-policy pilot concepts across accessibility, feasibility, legitimacy, equity, administrative-burden reduction, durability, risk, evidence quality, stakeholder coverage, and participation quality.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
pilots_path = article_dir / 'data' / 'raw' / 'public_policy_pilots_raw.csv'
pilots = pd.read_csv(pilots_path)
pilots
```

```python
pilots['evidence_strength'] = (
    0.40 * pilots['evidence_quality'] +
    0.35 * pilots['stakeholder_coverage'] +
    0.25 * pilots['participation_quality']
)
pilots['policy_value'] = (
    0.20 * pilots['accessibility'] +
    0.16 * pilots['feasibility'] +
    0.16 * pilots['legitimacy'] +
    0.20 * pilots['equity'] +
    0.14 * pilots['burden_reduction'] +
    0.08 * pilots['durability'] -
    0.06 * pilots['risk']
)
pilots['evidence_adjusted_value'] = pilots['policy_value'] * (0.75 + 0.25 * pilots['evidence_strength'])
pilots.sort_values('policy_value', ascending=False)
```
