# Design Thinking and Systems Thinking Intervention Analysis

This notebook accompanies the article **Design Thinking and Systems Thinking**. It provides an exploratory workflow for comparing interventions across human-centered value, systemic leverage, feasibility, equity sensitivity, durability, risk, evidence strength, and context sensitivity.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
portfolio_path = article_dir / 'data' / 'raw' / 'system_intervention_portfolio_raw.csv'
portfolio = pd.read_csv(portfolio_path)
portfolio
```

```python
portfolio['evidence_strength'] = 0.55 * portfolio['evidence_quality'] + 0.45 * portfolio['stakeholder_coverage']
portfolio['system_design_value'] = (
    0.24 * portfolio['human_value'] +
    0.26 * portfolio['system_leverage'] +
    0.18 * portfolio['feasibility'] +
    0.14 * portfolio['equity_sensitivity'] +
    0.12 * portfolio['durability'] -
    0.06 * portfolio['risk']
)
portfolio['evidence_adjusted_value'] = portfolio['system_design_value'] * (0.75 + 0.25 * portfolio['evidence_strength'])
portfolio.sort_values('system_design_value', ascending=False)
```
