# Insight Generation Analysis Notebook

This notebook accompanies the article **Insight Generation in Design Thinking**. It provides an exploratory workflow for comparing candidate insights across pattern support, explanatory depth, opportunity value, interpretive risk, evidence quality, stakeholder diversity, and prototype testability.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
data_path = article_dir / 'data' / 'raw' / 'candidate_insights_raw.csv'
df = pd.read_csv(data_path)
df
```

```python
weights = {
    'pattern_support': 0.30,
    'explanatory_depth': 0.30,
    'opportunity_value': 0.25,
    'interpretive_risk': 0.15,
}

df['insight_value'] = (
    weights['pattern_support'] * df['pattern_support'] +
    weights['explanatory_depth'] * df['explanatory_depth'] +
    weights['opportunity_value'] * df['opportunity_value'] -
    weights['interpretive_risk'] * df['interpretive_risk']
)

df.sort_values('insight_value', ascending=False)
```
