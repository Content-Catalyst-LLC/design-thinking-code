# Human-Centered Design Analysis Notebook

This notebook accompanies the article **Human-Centered Problem Solving**. It provides an exploratory workflow for comparing design options across human benefit, usability, stakeholder fit, burden, and evidence confidence.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
data_path = article_dir / 'data' / 'raw' / 'human_centered_options_raw.csv'
df = pd.read_csv(data_path)
df
```

```python
weights = {
    'human_benefit': 0.30,
    'usability': 0.25,
    'stakeholder_fit': 0.30,
    'burden': 0.15,
}

df['hc_value'] = (
    weights['human_benefit'] * df['human_benefit'] +
    weights['usability'] * df['usability'] +
    weights['stakeholder_fit'] * df['stakeholder_fit'] -
    weights['burden'] * df['burden']
)

df.sort_values('hc_value', ascending=False)
```
