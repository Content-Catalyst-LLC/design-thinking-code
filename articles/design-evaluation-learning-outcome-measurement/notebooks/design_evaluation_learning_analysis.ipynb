# Design Evaluation, Learning, and Outcome Measurement

This notebook accompanies the article **Design Evaluation, Learning, and Outcome Measurement**. It provides an exploratory workflow for comparing design interventions across outcomes, burden, equity, trust, durability, evidence strength, and learning priority.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
portfolio_path = article_dir / 'data' / 'raw' / 'evaluation_portfolio_raw.csv'
portfolio = pd.read_csv(portfolio_path)
portfolio
```

```python
portfolio['penalty'] = 0.50 * portfolio['operational_cost'] + 0.50 * portfolio['residual_risk']
portfolio['evaluation_value'] = (
    0.24 * portfolio['outcome_improvement'] +
    0.20 * portfolio['burden_reduction'] +
    0.20 * portfolio['equity_performance'] +
    0.16 * portfolio['trust_improvement'] +
    0.14 * portfolio['durability'] -
    0.06 * portfolio['penalty']
)
portfolio['evidence_strength'] = (
    0.40 * portfolio['evidence_quality'] +
    0.35 * portfolio['stakeholder_coverage'] +
    0.25 * portfolio['method_triangulation'] -
    0.20 * portfolio['uncertainty']
).clip(0, 1)
portfolio.sort_values('evaluation_value', ascending=False)
```
