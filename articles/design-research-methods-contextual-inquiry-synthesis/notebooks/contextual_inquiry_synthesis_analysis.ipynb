# Contextual Inquiry and Synthesis Analysis Notebook

This notebook accompanies the article **Design Research Methods: Contextual Inquiry and Synthesis**. It provides an exploratory workflow for analyzing contextual inquiry evidence, theme confidence, stakeholder coverage, method triangulation, and validation priorities.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
evidence_path = article_dir / 'data' / 'raw' / 'contextual_inquiry_evidence_units_raw.csv'
evidence = pd.read_csv(evidence_path)
evidence.head()
```

```python
theme_summary = (
    evidence.groupby('primary_theme')
    .agg(
        evidence_units=('unit_id', 'count'),
        stakeholder_groups=('participant_group', 'nunique'),
        methods=('method', 'nunique'),
        mean_evidence_strength=('evidence_strength', 'mean'),
        mean_interpretive_risk=('interpretive_risk', 'mean')
    )
    .reset_index()
)

theme_summary['synthesis_confidence'] = (
    0.35 * theme_summary['mean_evidence_strength'] +
    0.25 * theme_summary['stakeholder_groups'] +
    0.25 * theme_summary['methods'] -
    0.15 * theme_summary['mean_interpretive_risk']
)

theme_summary.sort_values('synthesis_confidence', ascending=False)
```
