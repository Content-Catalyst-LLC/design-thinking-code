# Design Thinking and Strategy

This notebook accompanies the article **Design Thinking and Strategy**. It provides an exploratory workflow for strategic option scoring, portfolio balance, uncertainty, implementation readiness, ethics, public value, and learning.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
options = pd.read_csv(article_dir / 'data' / 'raw' / 'strategic_options_raw.csv')
options.head()
```

```python
options['strategic_score'] = (
    0.17 * options['desirability'] +
    0.13 * options['feasibility'] +
    0.13 * options['viability'] +
    0.16 * options['strategic_alignment'] +
    0.11 * options['ethical_quality'] +
    0.10 * options['learning_value'] +
    0.08 * options['public_value'] +
    0.05 * options['evidence_strength'] * 10 -
    0.03 * options['strategic_risk'] -
    0.02 * options['implementation_effort'] -
    0.01 * options['capability_gap'] -
    0.01 * options['time_to_learn']
)
options.sort_values('strategic_score', ascending=False)
```
