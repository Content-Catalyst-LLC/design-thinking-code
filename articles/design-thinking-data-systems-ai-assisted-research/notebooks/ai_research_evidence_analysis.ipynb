# Design Thinking, Data Systems, and AI-Assisted Research

This notebook accompanies the article **Design Thinking, Data Systems, and AI-Assisted Research**. It explores research signal confidence, bias risk, AI-assistance risk, metadata quality, and decision readiness.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
signals = pd.read_csv(article_dir / 'data' / 'raw' / 'research_signals_raw.csv')
signals.head()
```

```python
signals['confidence_score'] = (
    0.18 * signals['source_strength'] +
    0.17 * signals['relevance'] +
    0.15 * signals['traceability'] +
    0.15 * signals['representativeness'] +
    0.15 * signals['validation'] +
    0.08 * signals['recency'] +
    0.07 * signals['consent_alignment'] +
    0.05 * signals['participant_coverage']
)
signals[['signal', 'evidence_type', 'confidence_score']].sort_values('confidence_score', ascending=False)
```
