# Co-Design and Participatory Design

This notebook accompanies the article **Co-Design and Participatory Design**. It provides an exploratory workflow for comparing participatory design activities across representation, accessibility, participant influence, trust, evidence quality, implementation accountability, decision impact, and ethical risk.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
activities_path = article_dir / 'data' / 'raw' / 'codesign_activities_raw.csv'
activities = pd.read_csv(activities_path)
activities
```

```python
activities['participation_quality'] = (
    0.18 * activities['representation'] +
    0.14 * activities['accessibility'] +
    0.22 * activities['participant_influence'] +
    0.12 * activities['trust_quality'] +
    0.12 * activities['evidence_quality'] +
    0.12 * activities['implementation_accountability'] +
    0.14 * activities['decision_impact'] -
    0.08 * activities['ethical_risk']
)
activities.sort_values('participation_quality', ascending=False)
```
