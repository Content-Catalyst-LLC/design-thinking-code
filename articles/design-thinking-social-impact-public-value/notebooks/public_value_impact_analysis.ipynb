# Design Thinking for Social Impact and Public Value

This notebook accompanies the article **Design Thinking for Social Impact and Public Value**. It explores social impact interventions, public value, impact readiness, stewardship need, equity, burden, participation, and portfolio priority.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
interventions = pd.read_csv(article_dir / 'data' / 'raw' / 'social_impact_interventions_raw.csv')
interventions.head()
```

```python
interventions['public_value_score'] = (
    0.13 * interventions['access'] +
    0.15 * interventions['equity'] +
    0.12 * interventions['dignity'] +
    0.12 * interventions['legitimacy'] +
    0.13 * interventions['accountability'] +
    0.11 * interventions['outcome_strength'] +
    0.08 * interventions['sustainability'] +
    0.07 * interventions['learning_capacity'] +
    0.05 * interventions['community_defined_value'] +
    0.04 * interventions['repair_capacity']
)
interventions[['intervention', 'intervention_type', 'public_value_score']].sort_values('public_value_score', ascending=False)
```
