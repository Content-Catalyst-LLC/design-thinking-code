# Design Thinking and Organizational Innovation

This notebook accompanies the article **Design Thinking and Organizational Innovation**. It provides an exploratory workflow for comparing innovation concepts across desirability, feasibility, viability, equity, learning value, implementation readiness, risk, evidence quality, and stakeholder coverage.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
concepts_path = article_dir / 'data' / 'raw' / 'organizational_innovation_concepts_raw.csv'
concepts = pd.read_csv(concepts_path)
concepts
```

```python
concepts['evidence_strength'] = (
    0.55 * concepts['evidence_quality'] +
    0.45 * concepts['stakeholder_coverage']
)
concepts['design_value'] = (
    0.22 * concepts['desirability'] +
    0.16 * concepts['feasibility'] +
    0.16 * concepts['viability'] +
    0.18 * concepts['equity'] +
    0.12 * concepts['learning_value'] +
    0.10 * concepts['implementation_readiness'] -
    0.06 * concepts['risk']
)
concepts['evidence_adjusted_value'] = concepts['design_value'] * (0.75 + 0.25 * concepts['evidence_strength'])
concepts.sort_values('design_value', ascending=False)
```
