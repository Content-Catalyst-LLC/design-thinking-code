# Service Design in Design Thinking

This notebook accompanies the article **Service Design in Design Thinking**. It provides an exploratory workflow for service-stage quality, end-to-end reliability, burden, accessibility, trust, recovery quality, and redesign priority.

```python
from pathlib import Path
import numpy as np
import pandas as pd

article_dir = Path('..')
stages_path = article_dir / 'data' / 'raw' / 'service_journey_stages_raw.csv'
stages = pd.read_csv(stages_path)
stages
```

```python
stages['service_stage_quality'] = (
    0.22 * stages['completion_probability'] * 10 +
    0.18 * stages['clarity'] +
    0.18 * stages['trust'] +
    0.16 * stages['accessibility'] +
    0.14 * stages['recovery_quality'] -
    0.07 * stages['user_burden'] -
    0.05 * stages['staff_load']
)
stages['end_to_end_reliability'] = np.prod(stages['completion_probability'])
stages.sort_values('service_stage_quality', ascending=False)
```
