# Ethics, Power, and Inclusion in Design Thinking

This notebook accompanies the article **Ethics, Power, and Inclusion in Design Thinking**. It provides an exploratory workflow for inclusion, burden, power gaps, ethical risk, governance need, accessibility, accountability, and repair.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
stakeholders = pd.read_csv(article_dir / 'data' / 'raw' / 'stakeholder_groups_raw.csv')
decisions = pd.read_csv(article_dir / 'data' / 'raw' / 'ethical_design_decisions_raw.csv')
stakeholders.head(), decisions.head()
```

```python
stakeholders['inclusion_score'] = (
    0.20 * stakeholders['access'] +
    0.20 * stakeholders['voice'] +
    0.16 * stakeholders['safety'] +
    0.14 * stakeholders['compensation'] +
    0.14 * stakeholders['representation'] +
    0.16 * stakeholders['accountability']
)
stakeholders[['group', 'inclusion_score']].sort_values('inclusion_score')
```
