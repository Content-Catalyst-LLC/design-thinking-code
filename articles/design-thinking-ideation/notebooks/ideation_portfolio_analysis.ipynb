# Ideation Portfolio Analysis Notebook

This notebook accompanies the article **Ideation in Design Thinking**. It provides an exploratory workflow for comparing candidate ideas across desirability, feasibility, novelty, equity value, learning value, residual risk, evidence quality, and prototype testability.

```python
from pathlib import Path
import pandas as pd

article_dir = Path('..')
data_path = article_dir / 'data' / 'raw' / 'idea_portfolio_raw.csv'
df = pd.read_csv(data_path)
df
```

```python
weights = {
    'desirability': 0.24,
    'feasibility': 0.18,
    'novelty': 0.18,
    'equity_value': 0.18,
    'learning_value': 0.12,
    'composite_risk': 0.10,
}

df['composite_risk'] = (
    0.25 * df['residual_risk'] +
    0.25 * df['ethical_risk'] +
    0.20 * df['operational_risk'] +
    0.15 * df['technical_risk'] +
    0.15 * df['scaling_risk']
)

df['idea_value'] = (
    weights['desirability'] * df['desirability'] +
    weights['feasibility'] * df['feasibility'] +
    weights['novelty'] * df['novelty'] +
    weights['equity_value'] * df['equity_value'] +
    weights['learning_value'] * df['learning_value'] -
    weights['composite_risk'] * df['composite_risk']
)

df.sort_values('idea_value', ascending=False)
```
